## Question 1:

You have been given two sequences that should be one, `seq1` and `seq2`. Both sequences will need to cleaned and then joined together, write some Python code that achieves the following goals:
- Remove the sequence ID from `seq1` (i.e. the characters before and including the `_`), then print the result
- `seq2` is missing some information, replace the underscores with `G`, then print the result
- Join the two sequences together, with `seq1` being first, then print the result

````python
seq1 = "SEQ001_ATCGATCGTAGCTAGCTAGCTA"
seq2 = "ATCG_CGTAGCTAG_CTAGCTA"
````

````python
seq1_cleaned = seq1.split("_", 1)[1]
print(seq1_cleaned)
````

````python
seq2_cleaned = seq2.replace("_", "G")
print(seq2_cleaned)
````

````python
combined_sequence = seq1_cleaned + seq2_cleaned
print(combined_sequence)
````

## Question 8:

You want to compare what blood types are unique or shared across group_a and group_b, the best way to do this is through sets:
- Create sets called `unique_group_a` and `unique_group_b` from the given lists
- Print the sets

````python
group_a_blood_types = ['A+', 'B+', 'O-', 'A+', 'AB+', 'O-', 'B+', 'A+', 'O+']
group_b_blood_types = ['O+', 'A-', 'B+', 'O+', 'A+', 'AB-', 'O-', 'A-']
````

````python
unique_group_a = set(group_a_blood_types)
unique_group_b = set(group_b_blood_types)

print(unique_group_a)
print(unique_group_b)
````

1. Find all blood types present across both groups

````python
all_blood_types = unique_group_a.union(unique_group_b)
print(all_blood_types)
````

2. Find blood types common to both groups

````python
shared_blood_types = unique_group_a.intersection(unique_group_b)
print(shared_blood_types)
````

3. Find blood types that appear in only **group B**, not **group A**

````python
group_b_only = unique_group_b.difference(unique_group_a)
print(group_b_only)
````
