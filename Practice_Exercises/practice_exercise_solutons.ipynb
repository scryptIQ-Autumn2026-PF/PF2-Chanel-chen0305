## Practice Exercise 14: Set Operations

You're analysing antimicrobial resistance patterns in bacterial isolates. You have resistance genes found in two different patient populations:

`population_A_resistance = {'blaTEM', 'aac', 'tetA', 'sul1', 'qnrS'}`
`population_B_resistance = {'blaCTX', 'aac', 'tetA', 'sul2', 'qnrS', 'mecA'}`

1. Find all resistance genes present across both populations
2. Find resistance genes common to both populations  
3. Find resistance genes unique to population A
4. Find resistance genes that appear in only one population (not both)

```python
population_A_resistance = {'blaTEM', 'aac', 'tetA', 'sul1', 'qnrS'}
population_B_resistance = {'blaCTX', 'aac', 'tetA', 'sul2', 'qnrS', 'mecA'}

# 1. All resistance genes across both populations (union)
all_genes = population_A_resistance.union(population_B_resistance)
# Alternative: all_genes = population_A_resistance | population_B_resistance
print("All resistance genes:", all_genes)

# 2. Resistance genes common to both (intersection)
common_genes = population_A_resistance.intersection(population_B_resistance)
# Alternative: common_genes = population_A_resistance & population_B_resistance
print("Common resistance genes:", common_genes)

# 3. Resistance genes unique to population A (difference)
unique_to_A = population_A_resistance.difference(population_B_resistance)
# Alternative: unique_to_A = population_A_resistance - population_B_resistance
print("Unique to population A:", unique_to_A)

# 4. Genes in only one population (symmetric difference)
exclusive_genes = population_A_resistance.symmetric_difference(population_B_resistance)
# Alternative: exclusive_genes = population_A_resistance ^ population_B_resistance
print("Genes in only one population:", exclusive_genes)
```

## Practice Exercise 21: String Methods

You're cleaning up a messy dataset of gene names and sample IDs from different laboratories. Each lab has used different formatting conventions that need to be standardised. Using the provided data:

1. Remove any leading/trailing whitespace from the gene name
2. Convert the gene name to uppercase for consistency
3. Replace any hyphens (-) with underscores (_) in the gene names
4. For the sample ID, split it at the underscore and create a formatted ID using f-strings: f"LAB-{lab_code}_SAMPLE-{number}"
5. Create a final summary string that combines the cleaned gene name and formatted sample ID

```python
# Messy data
gene_name = "  brca-1  "          
sample_id = "gilestro_024"

# 1. Remove whitespace
gene_name_clean = gene_name.strip()
print("After removing whitespace:", gene_name_clean)

# 2. Convert to uppercase
gene_name_upper = gene_name_clean.upper()
print("After converting to uppercase:", gene_name_upper)

# 3. Replace hyphens with underscores
gene_name_final = gene_name_upper.replace("-", "_")
print("After replacing hyphens:", gene_name_final)

# 4. Format the sample ID
sample_parts = sample_id.split("_")
lab_code = sample_parts[0]
number = sample_parts[1]
sample_id_formatted = f"LAB-{lab_code}_SAMPLE-{number}"
print("Formatted sample ID:", sample_id_formatted)

# 5. Create final summary
summary = f"Gene: {gene_name_final}, Sample: {sample_id_formatted}"
print("\nFinal summary:", summary)
```

## Practice Exercise 22: Converting Strings to Lists

- Use a method to achieve this aim
- Do not type "L" directly as an argument; use indexing to access "L" from within the tuple nested in the list

## Practice Exercise 23: Converting Strings to Tuples

Create a list from the following restriction enzyme recognition sequence and use it to check if it's *palindromic*.
1. Convert the string to a list
2. Use the `.reverse()` method to reverse the order
3. Check the equivalence of the reversed list to the original

*Hint: `.reverse()` acts in place, therefore you will need to make adjustments for this*

```python
restriction_site = "GAATTC"

# 1. Convert string to a list
site_list = list(restriction_site)
print("Original sequence as list:", site_list)

# 2. Create reversed list
reversed_list = site_list.copy()
reversed_list.reverse()
print("Reversed sequence:", reversed_list)

# 3. Check equivalence
is_palindrome = site_list == reversed_list
print(f"\nIs the sequence palindromic? {is_palindrome}")
```
